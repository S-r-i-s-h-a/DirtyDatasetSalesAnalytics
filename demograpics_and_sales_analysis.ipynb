31.Create a pivot table of TotalAmount by Region and Category.

```python
piv = pd.pivot_table(
    df[df['Returned'] == False],
    index=['Region', 'Category'],
    values='TotalAmount',
    aggfunc=['sum', 'mean'],
    margins=True,
    margins_name='Grand Total'
).round(2)
piv
```

32.Unpivot (melt) the pivot table from Task 31.


```python
piv2 = piv.reset_index()
piv2.index
```

```python
piv2.columns = [
    col[0] if col[1] == '' else f"{col[0]}_{col[1]}"
    for col in piv2.columns
]
print(piv2.columns)
```

```python
melted = piv2.melt(
    id_vars=['Region', 'Category'],
    value_vars=['sum_TotalAmount', 'mean_TotalAmount'],
    var_name='Metric',
    value_name='TotalAmountValue'
)
```

```python
print(melted)
```

39.Calculate return rate % by region (total returned orders / total orders).

```python
returned_by_reg = df.groupby('Region').agg(
    TotalOrders=('OrderID', 'count'),
    ReturnedOrders=('Returned', lambda x: x.sum())
)
returned_by_reg['Return Rate'] = returned_by_reg['ReturnedOrders'] /returned_by_reg['TotalOrders'] *100
```
